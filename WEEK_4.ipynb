```python
print("Ayush Prajapati 1WA24CS075")
from itertools import combinations

def calculate_cost(state):
    """Count pairs of queens attacking each other."""
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if (state[i] == state[j] or
                    abs(state[i] - state[j]) == abs(i - j)):
                conflicts += 1

    return conflicts


def hill_climbing(initial_state):
    current = initial_state[:]
    step = 0

    while True:
        current_cost = calculate_cost(current)

        print(f"\nStep {step}")
        print("Current state:", current)
        print("Current cost:", current_cost)

        neighbors = []

        # Generate all neighbors by swapping two queens
        for i, j in combinations(range(len(current)), 2):
            neighbor = current[:]
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            cost = calculate_cost(neighbor)
            neighbors.append((cost, i, j, neighbor))

            print(
                f"Swap (x{i}, x{j}): "
                f"{neighbor}, Cost = {cost}"
            )

        # Lowest cost first; ties use smallest index pair
        best = min(neighbors, key=lambda item: (
            item[0], item[1], item[2]
        ))

        best_cost, i, j, best_state = best

        print("\nBest neighbor:", best_state)
        print("Best neighbor cost:", best_cost)

        # Stop if no improvement is possible
        if best_cost >= current_cost:
            print("Terminated: no better neighbor.")
            print("Final state:", current)
            print("Final cost:", current_cost)
            return current

        current = best_state
        print(f"Selected swap: (x{i}, x{j})")
        step += 1

        if best_cost == 0:
            print("\nGoal state reached!")
            print("Final state:", current)
            print("Final cost:", best_cost)
            return current


# Given initial state
initial_state = [3, 1, 2, 0]

hill_climbing(initial_state)
```
